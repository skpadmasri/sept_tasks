```python
status = data['Booking Status'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(status.index, status.values, color='lightgreen', edgecolor='black')
plt.title('Booking Status Analysis')
plt.xlabel('Booking Status')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.show()
```

```python
payment = data['Payment Method'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(payment.index, payment.values, color='orange', edgecolor='black')
plt.title('Payment Method Analysis')
plt.xlabel('Payment Method')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.show()
```

```python
ratings = data['Customer Rating'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(ratings.index.astype(str), ratings.values, color='purple', edgecolor='black')
plt.title('Customer Ratings Analysis')
plt.xlabel('Rating')
plt.ylabel('Number of Customers')
plt.show()
```
